# Data load and feature data.

```python
# Importing dependencies
import pandas as pd
# MYSQL toolkit
import pymysql # work like SQL adapter
from sqlalchemy import create_engine 
```

```python
df = pd.read_csv('Walmart.csv',encoding_errors = 'ignore')
```

```python
df.head()
```

```python
df.describe()
```

```python
df.info()
```

```python
# To find value of duplicate
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace= True)
df.duplicated().sum()
```

```python
df.isnull().sum()
```

```python
# Dropping all rows with missing values
df.dropna(inplace=True)
df.isnull().sum()
```

```python
df.shape
```

```python
df['unit_price'] = df['unit_price'].str.replace('$','').astype(float)
```

```python
df.head()
```

```python
df.info()
```

```python
df.columns
```

```python
df['total'] = df['unit_price']*df['quantity']
df.head()
```

```python
df.shape
```

```python
df.to_csv('walmart_clean_data.csv',index=False)
```

```python
engine_mysql = create_engine("mysql+pymysql://root:xxxxxxxxxx@localhost:3306/walmart_db")
try:
    engine_mysql
    print("Connected with SQL")
except:
    print("Unable to connect")
```

```python
df.to_sql(name='walmart',con=engine_mysql,if_exists='append',index=False)
```
